# How to use the code

## Via this notebook:

### 1. Import library

```python
import source.musical_bingo_lib as mb
```

### 2. Set the parameters
These values will be needed:
- Total number of songs that will be used.
- Number of songs per card. It has to be a **multiple of 3**.
- Number of playes (number of cards that will be created).
- Number of games. For now, it can be either 1 or 2.

For example:


```python
total_number_of_songs = 36
songs_per_card = 12
number_of_cards = 60
number_of_games = 1
n_trials = 3e2  # optional
```

### 3. Simulate the kind of game we will have with these parameters

With function `mb.create_bingo_histogram`, we will calculate the average number of songs we will need to play for someone to achieve Bingo in a game, along with the histogram showing the average number of appearances per song. `n_trials` repetitions will be calculated in order to obtain the average.


**How to use it?** We will simply introduce the values we previously set (except for the number of games; as it will be the same for both games, unless we change a parameter).

Apart from that, we need to specify whether we want to plot the histogram, and whether we want to normalise the histogram (if we do so, we will lose information on the number of appearances per song). Set these values to True or False, as you wish.

```python
histogram = mb.create_bingo_histogram(total_number_of_songs = total_number_of_songs,   # you can modify these
                                       number_of_cards = number_of_cards,              # variables as you wish
                                       songs_per_card = songs_per_card,                # either here directly,
                                       _plot = False,                                  # or by modifying the
                                       normalise = False                               # varibale previously set
)
```

The average number of songs we will need for bingo was also printed. The histogram we created will be stored in variable `histogram`.


We did not plot it this time, but we can use function `mb.show_histogram` to do so:

```python
mb.show_histogram (histogram, total_number_of_songs=total_number_of_songs, number_of_cards=number_of_cards)
```

If we want to save the plot to a file, we just need to add the filename (path) in argument `save_image`:

```python
filename = "bingo_histogram.png"
mb.show_histogram(histogram, total_number_of_songs=total_number_of_songs, number_of_cards=number_of_cards, save_image=filename)
```

Now that we have this information, we can estimate how long it will take to finish one game, we just need to multiply the number of seconds we will play each song for with the average we just obtained (27.01).

### 3. Generate the cards

In order to generate the actual cards for the Bingo game, you will need an additional file, containing the list of songs that will be played. In directory `files/` you will find `example_list_1game.txt` and `example_list_2games.txt`, that will show you how to create lists for Bingo games with 1 game or 2 games.

As you can see, the only difference is that, if we want 2 games, we will need to specify the index of the game before each song name.

We will also need another variable, that contains the name of the file we will create with the LaTeX code for the cards. **Be careful**, if another file with the same name exists, it will be overwritten and you won't be able to recover the information.

In this section we will use the example files to generate the cards. To create your own, you only need to provide a song list and modify the parameters!

```python
# Number of songs that we will have per card (multiple of 3 !)
songs_per_card = 12

# Number of cards, depending of the number of playes
number_of_cards = 60

# List of songs
song_filename = "files/example_list_1game.txt"

# File to store the LaTeX code we will generate
latex_file = "whole_latex_code.txt"  # if you use a .tex extension, you can directly compile the file

# Generate the code:
mb.write_cards(songs_per_card,
               number_of_cards,
               song_filename,
               latex_file,
               add_colour = True,           # This is optional. Remove if you want it in black and white
               title = "Musical bingo!"     # This is optional. Default value = "MUSICAL BINGO"
)
```

### 4. Compile the code

Finally, you need to compile the code you just generated. In platform [Overleaf](https://www.overleaf.com/) it can be done really easily: create an account if you don't have one already, copy the code to a `.tex` file (or directly upload the file if you created a `.tex` file), and compile it. You will directly obtain the PDF with the cards.
